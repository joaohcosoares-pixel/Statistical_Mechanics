# 🧬 Simulação de Tensão Entrópica em Polímeros
### Modelo de Cadeia Livremente Articulada (Freely Jointed Chain - FJC 1D)

---

## 📌 1. Fundamentação Física e Termodinâmica

A elasticidade observada em materiais elastoméricos (como borracha natural) e biopolímeros (como filamentos de DNA, RNA e a proteína titina) é de **natureza puramente entrópica**.

Diferentemente de uma mola metálica ideal, onde a força restauradora decorre do aumento da **energia potencial intermolecular** ($U$) quando os átomos são afastados de sua posição de equilíbrio, um polímero ideal possui energia interna constante ($dU \approx 0$) durante a deformação conformacional.

A força elástica de restauração surge como uma consequência direta da **Segunda Lei da Termodinâmica**:

$$\mathrm{d}F = \mathrm{d}U - T\mathrm{d}S = -S\mathrm{d}T + J\mathrm{d}X$$

Para um processo isotérmico ($dT = 0$) em um polímero ideal ($dU = 0$):

$$J = -T \left( \frac{\partial S}{\partial X} \right)_T$$

---

## 📐 2. Formulação Matemática do Modelo 1D

1. **Cadeia com $N$ Elos de Comprimento $\ell$:**
   - Cada elo pode apontar para a **Direita** ($+1$) com número $n_R$, ou para a **Esquerda** ($-1$) com número $n_L = N - n_R$.
   - **Extensão Ponta-a-Ponta Líquida ($X$):**
     $$X = (n_R - n_L)\ell = (2n_R - N)\ell$$
   - **Extensão Fracionária ($x$):**
     $$x = \frac{X}{N\ell} = \frac{2n_R - N}{N} \in (-1, 1)$$

2. **Multiplicidade de Microestados ($\Omega$):**
   $$\Omega(n_R) = \binom{N}{n_R} = \frac{N!}{n_R!(N - n_R)!}$$

3. **Entropia Configuracional ($S$ via Stirling):**
   $$S(n_R) = k_B \ln \Omega(n_R) \approx k_B \big[ N\ln N - n_R\ln n_R - (N - n_R)\ln(N - n_R) \big]$$
   - O estado de **máxima desordem/entropia** ocorre em $n_R = N/2$ ($X = 0$), onde $S_{\max} = N k_B \ln 2$.
   - Ao esticar o polímero ($n_R \to N$), o número de configurações possíveis colapsa para $\Omega = 1$, de modo que $S \to 0$.

4. **Tensão Entrópica Exata ($J$):**
   $$J = -T \frac{\mathrm{d}S}{\mathrm{d}X} = -T \frac{1}{2\ell} \frac{\mathrm{d}S}{\mathrm{d}n_R} = \frac{k_B T}{2\ell} \ln\left( \frac{n_R}{N - n_R} \right) = \frac{k_B T}{\ell} \operatorname{arctanh}\left( \frac{X}{N\ell} \right)$$

5. **Limite Linear (Lei de Hooke Entrópica para $X \ll N\ell$):**
   Expandindo $\operatorname{arctanh}(x) \approx x + \mathcal{O}(x^3)$:
   $$J \approx \kappa X, \quad \text{onde} \quad \kappa = \frac{k_B T}{N\ell^2}$$

> **Propriedade Notável (Efeito Gough-Joule):** A constante de mola $\kappa \propto T$. Ao aquecer uma tira de borracha tracionada sob carga constante, ela **contrai**, pois a agitação térmica aumenta a tendência estatística ao enovelamento desordenado!

## 🔬 4. Análise e Conexão Física dos Fenômenos

1. **Conexão Microestado $\to$ Macroestado:**
   Ao clicar em *Nova Conformação*, a rota microscópica exata muda, porém as grandezas macroscópicas ($X, S, J$) permanecem idênticas. Isto reflete o postulado da física estatística em que o macroestado termodinâmico é governado pelo ensemble de microestados equiprováveis.

2. **Colapso da Entropia com o Estiramento:**
   O Gráfico 2 mostra que $S$ tem seu valor máximo no estado frouxo ($X=0$). Conforme $n_R \to N$, a multiplicidade $\Omega = \binom{N}{n_R}$ despenca de $\sim 10^{29}$ (para $N=100$) para exatamente $1$ no estado esticado ao limite, forçando a entropia a cair drasticamente.

3. **Surgimento e Divergência da Tensão Entrópica ($J$):**
   No Gráfico 3, a aproximação linear da Lei de Hooke (linha tracejada) descreve com precisão pequenas deformações ($|X| \le 0.3 N\ell$). Para extensões elevadas ($X \to N\ell$), a força real diverge assintoticamente ($J \to \infty$), expressando o limite físico de extensão da cadeia finita.
